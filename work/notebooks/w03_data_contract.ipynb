## 3. Verify it with queries (grain, counts, missing values, windows)

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Tal3at-M/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

print(f"Total Rows: {len(df):,}")
print(f"Columns: {list(df.columns)}")
print("\n--- Missing Values Check ---")
print(df.isnull().sum())

print("\n--- Target Distribution (trend_direction) ---")
print(df['trend_direction'].value_counts(normalize=True).round(4) * 100)

print("\n--- Summary Statistics ---")
print(df[['impressions_90d', 'avg_position', 'ctr', 'content_age_days']].describe().round(2))
```
