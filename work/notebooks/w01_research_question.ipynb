## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load dataset directly from raw GitHub URL
url = "https://raw.githubusercontent.com/Tal3at-M/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

total_pages = len(df)
declining_pages = (df["trend_direction"].str.lower() == "down").sum()
decline_rate = (declining_pages / total_pages) * 100
median_age = df["content_age_days"].median()

print(f"Total Pages Analyzed: {total_pages:,}")
print(f"Declining Pages: {declining_pages:,} ({decline_rate:.1f}%)")
print(f"Median Content Age: {median_age:.0f} days")
```
